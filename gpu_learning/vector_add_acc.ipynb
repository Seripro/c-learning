```python
%%writefile main.c
#include <stdio.h>
#include <stdlib.h>

int main(void) {
    long long N = 1000000;

    double *A = (double*)malloc(sizeof(double) * N);
    double *B = (double*)malloc(sizeof(double) * N);
    double *C = (double*)malloc(sizeof(double) * N);

    for (long long i = 0; i < N; i++) {
        A[i] = 1.0;
        B[i] = 2.0;
    }

    #pragma acc data copyin(A[0:N], B[0:N]) copyout(C[0:N])
    {
        #pragma acc kernels
        for (long long i = 0; i < N; i++) {
            C[i] = A[i] + B[i];
        }
    }

    printf("C[0] = %f\n", C[0]);

    free(A);
    free(B);
    free(C);

    return 0;
}
```
